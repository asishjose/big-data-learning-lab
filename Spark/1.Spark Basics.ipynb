```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("LearningSpark") \
    .getOrCreate()

print("Spark Started Successfully!")
```

```python
data = [
    (1, "Asish", 22, "ML"),
    (2, "Rahul", 24, "Data Science"),
    (3, "Anu", 21, "Web Dev"),
    (4, "Meera", 23, "ML")
]

columns = ["id", "name", "age", "domain"]

df = spark.createDataFrame(data, columns)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
# select specific columns

df.select("name", "domain").show()
```

```python
# filter rows

df.filter(df.age > 22).show()
```

```python
df.filter((df.age > 21) & (df.domain == "ML")).show()
```

```python
from pyspark.sql.functions import lit

df2 = df.withColumn("country", lit("India"))
df2.show()
```

```python
df3 = df.withColumnRenamed("domain", "specialization")
df3.show()
```

```python
df.orderBy("age").show()
```

```python
df.orderBy(df.age.desc()).show()
```

```python
df.groupby("domain").count().show()
```

```python
from pyspark.sql.functions import avg

df.select(avg("age")).show()
```

```python
df.createOrReplaceTempView("students")
```

```python
spark.sql("""
SELECT name, age
FROM students
WHERE age > 22
""").show()
```
