## Filtering

```python
df.filter(df['category'] == "Electronics").show()
```

```python
df.filter(df.price > 5000).show()
```

```python
df.filter(df.city == "Bangalore").show()
```

```python
df.filter(df.category == "Fashion").show()
```

```python
df.filter(df.quantity >= 2).show()
```

## Sorting

```python
df.orderBy("price").show()
```

```python
df.orderBy(df.quantity.desc()).show()
```

```python
df.sort(df.price.desc()).show(1)
```

## New Columns

```python
df = df.withColumn("total_price", df.price * df.quantity)

df.show(5)
```

```python
from pyspark.sql.functions import lit

df = df.withColumn("country", lit("India"))
```

```python
df.show(3)
```

## Aggregations

```python
from pyspark.sql.functions import sum

df.select(sum("total_price")).show()
```

```python
from pyspark.sql.functions import avg

df.select(avg("total_price")).show()
```

```python
df.columns
```

```python
from pyspark.sql.functions import max

df.select(max("price")).show()
```

```python
from pyspark.sql.functions import max

df.select(max("price").alias("max_price")).show()
```

```python
from pyspark.sql.functions import min

df.select(min("quantity").alias("Min Qty")).show()
```

## GroupBy

```python
df.groupBy("category").count().show()
```

```python
df.groupBy("city") \
    .agg(sum("total_price").alias("total_sales")) \
    .show()
```

```python
df.groupBy("category") \
    .agg(avg("price")) \
    .show()
```

```python
df.groupBy("category") \
    .avg("price") \
    .show()
```

```python
from pyspark.sql.functions import avg, max, min

df.groupBy("category").agg(
    avg("price").alias("avg_price"),
    max("price").alias("max_price"),
    min("price").alias("min_price")
).show()
```

```python
# total qty sold per product

df.groupBy("product") \
    .sum("quantity") \
    .show()
```
