```python
import pandas as pd
import random

# -----------------------------
# CONFIG
# -----------------------------
NUM_ROWS = 3000

data = []

# -----------------------------
# DATA GENERATION
# -----------------------------
for _ in range(NUM_ROWS):

    # Number of active orders
    order_count = random.randint(5, 100)

    # Rain condition (0 = No Rain, 1 = Rain)
    rain = random.randint(0, 1)

    # Delivery time logic
    base_time = random.randint(10, 30)

    # More orders increase delay
    delivery_time = base_time + (order_count * 0.4)

    # Rain increases delivery time
    if rain == 1:
        delivery_time += random.randint(5, 15)

    # Small randomness
    delivery_time += random.randint(-3, 5)

    # Round value
    delivery_time = round(delivery_time, 2)

    # Delay label logic
    if delivery_time > 40:
        delay = "Late"
    else:
        delay = "OnTime"

    data.append([
        order_count,
        rain,
        delivery_time,
        delay
    ])

# -----------------------------
# CREATE DATAFRAME
# -----------------------------
columns = [
    "order_count",
    "rain",
    "delivery_time",
    "delay"
]

df = pd.DataFrame(data, columns=columns)

# -----------------------------
# SAVE DATASET
# -----------------------------
df.to_csv("delivery_delay_dataset.csv", index=False)

# -----------------------------
# SAMPLE OUTPUT
# -----------------------------
print("Dataset created successfully!\n")
print(df.head())
```
